# TikZ Primitives

tikzfigure wraps common TikZ drawing primitives so you don't need raw LaTeX for
standard geometric shapes.

- arcs and circles
- rectangles and ellipses
- grids and parabolas
- lines and intersections
- regular polygons (triangle, square, pentagon, ...)

```python
import math

from tikzfigure import TikzFigure, arrows, styles
```

## Arcs

An arc is defined by a starting point, start and end angles (in degrees), and a
radius.

```python
fig = TikzFigure()

# Quarter circle arc
fig.arc(
    start=(0, 0),
    start_angle=0,
    end_angle=90,
    radius="2cm",
    color="red",
    line_width="2pt",
)

# Three-quarter circle arc
fig.arc(
    start=(4, 0),
    start_angle=90,
    end_angle=360,
    radius="1.5cm",
    color="blue",
    line_width="2pt",
)

# Arc with arrow
fig.arc(
    start=(8, 0),
    start_angle=0,
    end_angle=180,
    radius="2cm",
    arrows=arrows.forward,
    color="green",
    line_width="2pt",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Circles

A circle is defined by a center point and a radius.

```python
fig = TikzFigure()

# Unfilled circle
fig.circle(
    center=(0, 0),
    radius="1cm",
    color="blue",
    line_width="2pt",
)

# Filled circle
fig.circle(
    center=(3, 0),
    radius="1cm",
    fill="yellow",
    color="orange",
    line_width="2pt",
)

# Dashed circle
fig.circle(
    center=(6, 0),
    radius="1cm",
    color="purple",
    dash_pattern="on 2pt off 2pt",
    line_width="2pt",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Clock face

A practical example: a clock face using a circle and line segments for hour
markers.

```python
fig = TikzFigure()

# Outer circle
fig.circle(
    center=(0, 0),
    radius="3cm",
    color="black",
    line_width="2pt",
)

# Hour markers (12 arcs around the circle)
for hour in range(12):
    angle = ((hour + 3) * 30) % 360  # 30 degrees apart
    rad = math.radians(angle)

    # Small line segment from inner to outer
    x_inner = 2.7 * math.cos(rad)
    y_inner = 2.7 * math.sin(rad)
    fig.draw(
        [(x_inner, y_inner), (x_inner * 1.1, y_inner * 1.1)],
        line_width="1.5pt",
        color="black",
    )

    # Hour numbers
    x_text = 2.4 * math.cos(rad)
    y_text = 2.4 * math.sin(rad)
    fig.node(
        x_text,
        y_text,
        content=str(hour if hour > 0 else 12),
        draw="none",
    )

# Center dot
fig.circle(
    center=(0, 0),
    radius="0.15cm",
    fill="black",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Rectangles

Rectangles are defined by two opposite corners.

```python
fig = TikzFigure()

# Unfilled rectangle
fig.rectangle(
    (0, 0),
    (2, 1),
    color="blue",
    line_width="2pt",
)

# Filled rectangle
fig.rectangle(
    (3, 0),
    (5, 1),
    fill="yellow",
    color="orange",
    line_width="2pt",
)

# Rectangle with rounded corners
fig.rectangle(
    (6, 0),
    (8, 1),
    fill="green!30",
    color="green",
    rounded_corners="5pt",
    line_width="2pt",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Ellipses

Ellipses are defined by a center point and two radii.

```python
fig = TikzFigure()

# Circle (equal radii)
fig.ellipse(
    (0, 0),
    x_radius="1cm",
    y_radius="1cm",
    color="red",
    line_width="2pt",
)

# Horizontal ellipse
fig.ellipse(
    (4, 0),
    x_radius="1.5cm",
    y_radius="0.75cm",
    color="blue",
    line_width="2pt",
)

# Filled ellipse
fig.ellipse(
    (8, 0),
    x_radius="1.2cm",
    y_radius="0.8cm",
    fill="purple!40",
    color="purple",
    line_width="2pt",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Grids

A grid spans between two opposite corners with a specified step size.

```python
fig = TikzFigure()

# Uniform grid
fig.grid(
    (0, 0),
    (3, 3),
    step="0.5cm",
    color="gray!30",
)

# Grid with different x and y steps
fig.grid(
    (4, 0),
    (7, 3),
    xstep="0.5cm",
    ystep="1cm",
    color="blue!20",
)

# Grid with larger step
fig.grid(
    (8, 0),
    (10, 3),
    step="1cm",
    color="red!20",
    line_width="1pt",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Parabolas

Parabolas connect two points, optionally with a bend (control) point.

```python
fig = TikzFigure()

# Simple parabola
fig.parabola(
    (0, 0),
    (3, 0),
    color="red",
    line_width="2pt",
)

# Parabola with bend point (creates a specific curve shape)
fig.parabola(
    (5, 0),
    (8, 0),
    bend=(6.5, 2),
    color="blue",
    line_width="2pt",
)

# Parabola with arrow
fig.parabola(
    (0, 3),
    (3, 4),
    color="green",
    arrows=arrows.forward,
    line_width="2pt",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Lines and intersections

Lines connect two points and support arrows and all standard styling.

```python
fig = TikzFigure()

# Simple line
fig.line(
    (0, 0),
    (2, 0),
    color="red",
    line_width="2pt",
)

# Line with arrow
fig.line(
    (3, 0),
    (5, 2),
    color="blue",
    arrows=arrows.forward,
    line_width="2pt",
)

# Dashed line
fig.line(
    (6, 0),
    (8, 1),
    color="green",
    options=[styles.dashed],
    line_width="2pt",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

Find where two lines intersect using the `intersection()` method:

```python
fig = TikzFigure()

from tikzfigure.core.line import Line

# Create two lines
line1 = fig.line(
    (0, 0),
    (4, 4),
    color="red",
    line_width="2pt",
)
line2 = fig.line(
    (0, 4),
    (4, 0),
    color="blue",
    line_width="2pt",
)

# Find intersection
intersection = line1.intersection(line2)
print(f"Lines intersect at: {intersection}")

# Draw intersection point
if intersection:
    fig.node(
        intersection[0],
        intersection[1],
        shape="circle",
        fill="black",
        minimum_size="0.15cm",
        draw="none",
    )

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Regular polygons

Draw regular polygons (triangle, square, pentagon, etc.) with specified center
and radius.

```python
fig = TikzFigure()

# Triangle (3 sides)
fig.triangle(
    (0, 1),
    radius=0.8,
    color="red",
    fill="red!20",
)

# Square (4 sides)
fig.square(
    (3, 1),
    radius=0.8,
    color="blue",
    fill="blue!20",
)

# Pentagon (5 sides)
fig.polygon(
    (6, 1),
    radius=0.8,
    sides=5,
    color="green",
    fill="green!20",
)

# Hexagon (6 sides)
fig.polygon(
    (9, 1),
    radius=0.8,
    sides=6,
    color="orange",
    fill="orange!20",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Rotated polygons

Polygons can be rotated by specifying a rotation angle in degrees.

```python
fig = TikzFigure()

# Squares with different rotations
rotations = [0, 22.5, 45]
for i, rot in enumerate(rotations):
    fig.square(
        (i * 3, 1),
        radius=0.8,
        rotation=rot,
        color="purple",
        fill="purple!30",
    )
    fig.node(
        i * 3,
        -0.3,
        content=f"{rot}\\textdegree",
        draw="none",
    )

# Triangles pointing in different directions
fig.triangle(
    (0, 3.5),
    radius=0.8,
    rotation=0,
    color="blue",
    fill="blue!20",
)
fig.triangle(
    (3, 3.5),
    radius=0.8,
    rotation=120,
    color="blue",
    fill="blue!20",
)
fig.triangle(
    (6, 3.5),
    radius=0.8,
    rotation=240,
    color="blue",
    fill="blue!20",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Matrix of nodes

`add_matrix()` lays out a grid of nodes using TikZ's `matrix` library, so you
don't need to hand-write raw `\matrix` code. Cells are given row by row, and
per-cell styling is set with a `{"content": ..., **options}` dict. Use
`row_sep`, `column_sep`, and `cell_style` to control spacing and the shared
look of every cell.

```python
fig = TikzFigure()

fig.add_matrix(
    [
        ["A", "B", "C"],
        ["D", {"content": "E", "fill": "yellow!60"}, "F"],
    ],
    label="grid",
    options="matrix of nodes",
    row_sep="5pt",
    column_sep="5pt",
    cell_style="draw, minimum size=1cm, anchor=center",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

Individual cells are addressable with `Matrix.cell(row, col)`, which returns
the TikZ node name for that cell (`"<label>-<row>-<col>"`, 1-indexed). Turn
that name into a figure coordinate with `add_coordinate(..., at=...)` to draw
connections between cells:

```python
fig = TikzFigure()

grid = fig.add_matrix(
    [["A", "B"], ["C", "D"]],
    label="grid",
    options="matrix of nodes",
    row_sep="5pt",
    column_sep="5pt",
    cell_style="draw, minimum size=1cm, anchor=center",
)

fig.add_coordinate("c1", at=grid.cell(1, 1))
fig.add_coordinate("c2", at=grid.cell(2, 2))
fig.draw(["c1", "c2"], arrows=arrows.forward, color="red", line_width="1.5pt")

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Fit boxes around nodes

`add_fit()` wraps a group of existing nodes/coordinates in a single bounding
node, using TikZ's `fit` library - handy for drawing a background box around
related elements without hand-writing `fit=(...)...`. Pass the `Node`/
`Coordinate` objects (or their label strings) to fit around.

```python
fig = TikzFigure()

a = fig.node((0, 0), shape="circle", fill="blue!30", content="A")
b = fig.node((3, 0), shape="circle", fill="blue!30", content="B")
c = fig.node((1.5, 2), shape="circle", fill="blue!30", content="C")

fig.add_fit(
    [a, b, c],
    options=["draw", "dashed", "rounded corners"],
    inner_sep="0.4cm",
)

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

Fit nodes can be styled like any other node (`fill`, `color`, ...) and, since
`add_fit()` returns a regular `Node`-like object, a title can be placed above
it at an absolute coordinate:

```python
fig = TikzFigure()

x = fig.node((0, 0), content="x")
y = fig.node((2, 0), content="y")
z = fig.node((1, 1.5), content="z")

fig.add_fit(
    [x, y, z],
    label="group",
    options=["draw"],
    color="orange!60",
    fill="orange!10",
    inner_sep="0.3cm",
)
fig.node((1, 2.3), content="\\textbf{Group}", draw="none")

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```

## Circle of nodes

Placing many nodes evenly around a circle using a Python loop together with TikZ
variables (defined via `variable`).

```python
fig = TikzFigure()

fig.variable("radius", 3)

N = 12
for i in range(N):
    frac = round(i / N, 4)
    fig.node(
        (
            f"{{\\radius*{math.cos(frac * 2 * math.pi):.4f}}}",
            f"{{\\radius*{math.sin(frac * 2 * math.pi):.4f}}}",
        ),
        shape="circle",
        fill="blue!40",
        content=str(i),
        minimum_size="0.6cm",
    )

fig.show()
```

```python
print(fig)
```

```python
print(fig.generate_standalone())
```
